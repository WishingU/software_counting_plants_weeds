```python
from PIL import Image
import os

input_folder = r"path/to/input/images"
output_folder = r"path/to/output/images"

os.makedirs(output_folder, exist_ok=True)

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")

for filename in os.listdir(input_folder):

    if not filename.lower().endswith(image_extensions):
        continue

    input_path = os.path.join(input_folder, filename)

    img = Image.open(input_path)

    width, height = img.size

   
    x_points = [0, width // 3, (width * 2) // 3, width]
    y_points = [0, height // 3, (height * 2) // 3, height]

    base_name = os.path.splitext(filename)[0]

    part_number = 1

    for row in range(3):
        for col in range(3):

            left = x_points[col]
            upper = y_points[row]
            right = x_points[col + 1]
            lower = y_points[row + 1]

            cropped_img = img.crop((left, upper, right, lower))

            output_name = f"{base_name}_part{part_number}.jpg"
            output_path = os.path.join(output_folder, output_name)

            if cropped_img.mode != "RGB":
                cropped_img = cropped_img.convert("RGB")

            cropped_img.save(output_path, quality=95)

            print(f"Saved: {output_name}")

            part_number += 1

print("Finished! All images have been split into 9 parts.")
```
