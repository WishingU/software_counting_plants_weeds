```python
import os
from PIL import Image
from pillow_heif import register_heif_opener


register_heif_opener()

input_folder = r"path/to/input/images"
output_folder = r"path/to/output/images"


os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(input_folder):

    if filename.lower().endswith((".heic", ".jpg", ".jpeg", ".png")):

        image_path = os.path.join(input_folder, filename)


        img = Image.open(image_path).convert("RGB")

        width, height = img.size


        tile_width = width // 3
        tile_height = height // 3

        base_name = os.path.splitext(filename)[0]

        count = 1


        for row in range(3):
            for col in range(3):

                left = col * tile_width
                top = row * tile_height


                if col == 2:
                    right = width
                else:
                    right = (col + 1) * tile_width

                if row == 2:
                    bottom = height
                else:
                    bottom = (row + 1) * tile_height

                cropped = img.crop((left, top, right, bottom))

                # 保存为 jpg
                output_name = f"{base_name}_{count}.jpg"
                output_path = os.path.join(output_folder, output_name)

                cropped.save(output_path, quality=95)

                count += 1

        print(f"Finished: {filename}")

print("All images have been split!")
```
